```python
import matplotlib.pyplot as plt

category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))

category_sales.plot(kind="bar", color="skyblue")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)

plt.tight_layout()

# Save FIRST
plt.savefig("../charts/sales_by_category.png", dpi=300, bbox_inches="tight")

# Then display
plt.show()

# Close the figure
plt.close()
```

```python
category_profit = df.groupby("Category")["Profit"].sum()

plt.figure(figsize=(8,5))
category_profit.plot(kind="bar")

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.xticks(rotation=0)

plt.tight_layout()

# Save the chart
plt.savefig("../charts/profit_by_category.png", dpi=300, bbox_inches="tight")

# Display the chart
plt.show()

# Close the figure
plt.close()
```

# Business Insights

## Key Findings

- Technology category generated the highest sales.
- Office Supplies contributed consistent profits.
- The West region achieved the highest revenue.
- California was the highest-performing state by sales.
- Consumer segment generated the maximum sales.
- Higher discounts generally reduced profit margins.

## Business Recommendations

- Increase investment in high-performing categories.
- Reduce excessive discounts on low-profit products.
- Focus marketing campaigns in high-revenue regions.
- Improve sales strategies in underperforming states.
